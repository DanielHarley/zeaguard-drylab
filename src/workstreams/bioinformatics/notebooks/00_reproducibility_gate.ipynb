# Notebook 00: reproducibility gate

Verify the declared software environment and publish a local snapshot. The gate does not install dependencies or perform network checks.

```python
import zeaguard.repro as repro

ALLOW_DIRTY = False
```

```python
config, report, collected = repro.run_all_checks(allow_dirty=ALLOW_DIRTY)
context = collected["context"]
context
```

```python
print(report.to_text())
```

```python
report.raise_if_failed()
```

```python
snapshot = repro.build_snapshot(context, config, report, collected, allow_dirty=ALLOW_DIRTY)
snapshot_path = repro.write_env_snapshot(snapshot, config.snapshot_dir)
snapshot_path
```

Downstream notebooks must call `repro.assert_environment_ready()` before analysis.
